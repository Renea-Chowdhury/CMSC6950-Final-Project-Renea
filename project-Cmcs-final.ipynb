# Converting the "Date" coloumn to "Datetime" and find the range of all the quarters

```python


# Ensure the 'date' column is in datetime format
df['date'] = pd.to_datetime(df['date'])

# Define the function to find date range for a given quarter
def find_daterange(quarter):
    # Filter the dataset for the given quarter
    quarter_data = df[df['quarter'] == quarter]
    
    # Check if the quarter exists in the dataset
    if not quarter_data.empty:
        min_date = quarter_data['date'].min().strftime('%m/%d/%Y')
        max_date = quarter_data['date'].max().strftime('%m/%d/%Y')
        print(f"Dates between {min_date} and {max_date} in {quarter}")
    else:
        print(f"No data found for {quarter}")

# Call the function for each quarter
find_daterange('Quarter1')
find_daterange('Quarter2')
find_daterange('Quarter3')
find_daterange('Quarter4')
find_daterange('Quarter5')  # Quarter5 may not exist
```

```python
# Histograms of numerical variables

fig, axes = plt.subplots(nrows=4, ncols=2, figsize=(15, 15))

for i, ax in enumerate(axes.ravel()):
    column_name = con[i]

    # Skip if the column is missing or has no valid data
    if column_name not in df.columns or df[column_name].isnull().all():
        ax.set_title(f"{title(column_name)}\nNo Data Available")
        ax.set_axis_off()
        continue

    # Plot histogram
    n, bins, patches = ax.hist(df[column_name].dropna(), bins=10, color="skyblue", edgecolor="black")
    ax.set_title(f"Distribution of {title(column_name)}")
    ax.set_ylabel("Frequency")

    # Create DataFrame with bin data
    df_hist = pd.DataFrame({'bin_values': n, 'left_bin_edge': bins[:-1]})
    last = {'bin_values': 0, 'left_bin_edge': bins[-1]}
    df_hist = pd.concat([df_hist, pd.DataFrame([last])], ignore_index=True)

    # Find bin with the highest frequency
    max_bin = df_hist[df_hist['bin_values'] == df_hist['bin_values'].max()]
    if not max_bin.empty:
        left_edge = max_bin['left_bin_edge'].iloc[0]
        right_edge = bins[bins > left_edge][0]

        # Add annotation about the most frequent bin
        if column_name in ['no_of_workers', 'idle_men']:
            ax.set_xlabel(f"{title(column_name)}\nMost values fell in the {int(left_edge)} to {int(right_edge)} bin")
        else:
            ax.set_xlabel(f"{title(column_name)}\nMost values fell in the {round(left_edge, 2)} to {round(right_edge, 2)} bin")

# Adjust layout
fig.tight_layout()
plt.subplots_adjust(top=1.5)
plt.show()
```

# Bar graph of **Actual productivity of each team**

```python


# Calculate the average actual productivity for each team
avg_productivity_per_team = df.groupby('team')['actual_productivity'].mean().reset_index()

# Create the bar plot
plt.figure(figsize=(10, 6))
bar_plot = sns.barplot(
    x='team', 
    y='actual_productivity', 
    data=avg_productivity_per_team, 
    palette='viridis'
)

# Annotate the bars with the actual values
for bar in bar_plot.containers[0]:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # X position
        height,  # Y position (height of the bar)
        f'{height:.2f}',  # Text (formatted to 2 decimal places)
        ha='center',  # Horizontal alignment
        va='bottom',  # Vertical alignment
        fontsize=10,
        color='black'
    )

# Customize the plot
plt.title('Average Actual Productivity for Each Team', fontsize=16)
plt.xlabel('Team', fontsize=14)
plt.ylabel('Average Actual Productivity', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```

## Representation of *Actual and Targeted Productivity*

```python

# Calculate average actual and targeted productivity per team
df_garment_bar_actual_team = df.groupby(['team'])['actual_productivity'].mean()
df_garment_bar_target_team = df.groupby(['team'])['targeted_productivity'].mean()

# x axis (teams)
teams = df['team'].sort_values(ascending=True).unique().tolist()

# The x position of bars
barWidth = 0.3
r1 = np.arange(len(teams))  # position for target bars
r2 = [x + barWidth for x in r1]  # position for actual bars

# Plotting teams
plt.figure(figsize=(10, 5))

# Plot target productivity
bars_target = plt.bar(r1, df_garment_bar_target_team, width=barWidth, color='#ff9ae6', edgecolor='black', capsize=7, label='Target')

# Plot actual productivity
bars_actual = plt.bar(r2, df_garment_bar_actual_team, width=barWidth, color="#ff1095", edgecolor='black', capsize=7, label='Actual')

# Annotate target bars with values
for bar in bars_target:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # X position (middle of the bar)
        height,  # Y position (height of the bar)
        f'{height:.2f}',  # Text with formatted height
        ha='center',  # Horizontal alignment
        va='bottom',  # Vertical alignment (text will appear above the bar)
        fontsize=10,
        color='black'
    )

# Annotate actual bars with values
for bar in bars_actual:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # X position (middle of the bar)
        height,  # Y position (height of the bar)
        f'{height:.2f}',  # Text with formatted height
        ha='center',  # Horizontal alignment
        va='bottom',  # Vertical alignment (text will appear above the bar)
        fontsize=10,
        color='black'
    )

# General layout
plt.xticks(r2, labels=teams)
plt.ylim(0, 1)  # Adjust y-axis limit if needed
plt.xlabel('Team')
plt.ylabel('Productivity')
plt.title('Target and Actual Productivity of Teams')
plt.legend()

# Show the plot
plt.tight_layout()
plt.show()
```

# Bar plot for **Target achieved by Departments**

```python
# Count instances where productivity met or exceeded the target, grouped by department
df['achieved_target'] = df['actual_productivity'] >= df['targeted_productivity']
target_achieved_counts = df.groupby('department')['achieved_target'].sum().reset_index()

# Rename columns for clarity
target_achieved_counts.rename(columns={'achieved_target': 'Instances Achieved'}, inplace=True)

# Create the bar plot
plt.figure(figsize=(10, 6))
bar_plot = sns.barplot(
    x='department',
    y='Instances Achieved',
    data=target_achieved_counts,
    palette='viridis'
)

# Annotate the bars with actual values
for bar in bar_plot.containers[0]:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # X position
        height,  # Y position (height of the bar)
        f'{int(height)}',  # Actual value
        ha='center',  # Horizontal alignment
        va='bottom',  # Vertical alignment
        fontsize=10,
        color='black'
    )

# Customize the plot
plt.title('Number of Instances Where Target Was Achieved by Department', fontsize=16)
plt.xlabel('Department', fontsize=14)
plt.ylabel('Instances Achieved', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.show()
```

# Generate total productivity by month

```python
# Extract the month from the 'date' column
df['month'] = df['date'].dt.month

# Calculate total productivity by month
total_productivity_by_month = df.groupby('month')['actual_productivity'].sum().reset_index()

# Create a bar plot
plt.figure(figsize=(10, 6))
bar_plot = sns.barplot(
    x='month', 
    y='actual_productivity', 
    data=total_productivity_by_month, 
    palette='viridis'
)

# Annotate the bars with the actual values
for bar in bar_plot.containers[0]:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # X position
        height,  # Y position (height of the bar)
        f'{height:.2f}',  # Text with 2 decimal places
        ha='center',  # Horizontal alignment
        va='bottom',  # Vertical alignment
        fontsize=10,
        color='black'
    )

# Customize the plot
plt.title('Total Productivity by Month', fontsize=16)
plt.xlabel('Month', fontsize=14)
plt.ylabel('Total Productivity', fontsize=14)
plt.xticks(ticks=range(0, 12), labels=[
    'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
    'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'], fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```

# Generate bar graph for Average incentive and overtime for each team

```python
# Calculate the average incentive and overtime for each team
team_avg = df.groupby('team')[['incentive', 'over_time']].mean().reset_index()

# Melt the data for a grouped bar chart
team_avg_melted = team_avg.melt(id_vars='team', var_name='Metric', value_name='Average')

# Create a grouped bar plot
plt.figure(figsize=(12, 6))
bar_plot = sns.barplot(
    x='team', 
    y='Average', 
    hue='Metric', 
    data=team_avg_melted, 
    palette='viridis'
)

# Annotate the bars with the actual values
for bar in bar_plot.containers:
    for b in bar:
        height = b.get_height()
        if height > 0:  # Avoid annotating for zero-height bars
            bar_plot.text(
                b.get_x() + b.get_width() / 2,  # X position
                height,  # Y position (height of the bar)
                f'{height:.2f}',  # Text with 2 decimal places
                ha='center',  # Horizontal alignment
                va='bottom',  # Vertical alignment
                fontsize=9,
                color='black'
            )

# Customize the plot
plt.title('Average Incentive and Overtime by Team', fontsize=16)
plt.xlabel('Team', fontsize=14)
plt.ylabel('Average Value', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(title='Metric', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```
