```python
import csv
import matplotlib.pyplot as plt
from datetime import datetime

# Step 1: Load the CSV data manually without pandas
file_path = r"C:\Users\salek\garments_worker_productivity.csv"
  # Update with your actual file path

# Initialize empty lists to store the 'date' and 'actual_productivity' values
dates = []
productivity = []

# Step 2: Read the CSV file from the folder
try:
    with open(file_path, mode='r') as file:
        csv_reader = csv.DictReader(file)
        
        for row in csv_reader:
            # Convert the 'date' column to datetime and store in the list
            try:
                date = datetime.strptime(row['date'], "%m/%d/%Y")  # Adjust the format if needed
                actual_productivity = float(row['actual_productivity'])
                dates.append(date)
                productivity.append(actual_productivity)
            except ValueError:
                # Handle any rows where conversion fails (e.g., if there's invalid data)
                continue

    # Step 3: Plot the data
    plt.figure(figsize=(12, 6))
    plt.plot(dates, productivity, marker='o', linestyle='-', color='b')
    plt.title('Actual Productivity Over Time', fontsize=16)
    plt.xlabel('Date', fontsize=14)
    plt.ylabel('Actual Productivity', fontsize=14)
    plt.xticks(rotation=45)
    plt.grid(True)
    plt.tight_layout()

    # Display the plot
    plt.show()
    
except FileNotFoundError as e:
    print(f"File not found: {e}")
```
