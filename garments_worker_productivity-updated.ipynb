# Generate total productivity by month

```python
# Extract the month from the 'date' column
df['month'] = df['date'].dt.month

# Calculate total productivity by month
total_productivity_by_month = df.groupby('month')['actual_productivity'].sum().reset_index()

# Create a bar plot
plt.figure(figsize=(10, 6))
bar_plot = sns.barplot(
    x='month', 
    y='actual_productivity', 
    data=total_productivity_by_month, 
    palette='viridis'
)

# Annotate the bars with the actual values
for bar in bar_plot.containers[0]:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # X position
        height,  # Y position (height of the bar)
        f'{height:.2f}',  # Text with 2 decimal places
        ha='center',  # Horizontal alignment
        va='bottom',  # Vertical alignment
        fontsize=10,
        color='black'
    )

# Customize the plot
plt.title('Total Productivity by Month', fontsize=16)
plt.xlabel('Month', fontsize=14)
plt.ylabel('Total Productivity', fontsize=14)
plt.xticks(ticks=range(0, 12), labels=[
    'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
    'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'], fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```

# Generate bar graph for Average incentive and overtime for each team

```python
# Calculate the average incentive and overtime for each team
team_avg = df.groupby('team')[['incentive', 'over_time']].mean().reset_index()

# Melt the data for a grouped bar chart
team_avg_melted = team_avg.melt(id_vars='team', var_name='Metric', value_name='Average')

# Create a grouped bar plot
plt.figure(figsize=(12, 6))
bar_plot = sns.barplot(
    x='team', 
    y='Average', 
    hue='Metric', 
    data=team_avg_melted, 
    palette='viridis'
)

# Annotate the bars with the actual values
for bar in bar_plot.containers:
    for b in bar:
        height = b.get_height()
        if height > 0:  # Avoid annotating for zero-height bars
            bar_plot.text(
                b.get_x() + b.get_width() / 2,  # X position
                height,  # Y position (height of the bar)
                f'{height:.2f}',  # Text with 2 decimal places
                ha='center',  # Horizontal alignment
                va='bottom',  # Vertical alignment
                fontsize=9,
                color='black'
            )

# Customize the plot
plt.title('Average Incentive and Overtime by Team', fontsize=16)
plt.xlabel('Team', fontsize=14)
plt.ylabel('Average Value', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(title='Metric', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```

# Generate result for Target achieve rate and number of workers

```python
# Step 1: Replace inf and NaN values in achievement_rate column
df['achievement_rate'] = (df['actual_productivity'] / df['targeted_productivity']) * 100
df['achievement_rate'].replace([np.inf, -np.inf], np.nan, inplace=True)

# Step 2: Drop rows where achievement_rate is NaN (optional, based on your needs)
df.dropna(subset=['achievement_rate'], inplace=True)

# Step 3: Group and Calculate Metrics
team_metrics = df.groupby('team').agg({
    'achievement_rate': 'mean',
    'no_of_workers': 'sum'
}).reset_index()

# Step 4: Visualization
fig, ax1 = plt.subplots(figsize=(12, 6))

# Bar plot for achievement rate
sns.barplot(
    x='team', y='achievement_rate', data=team_metrics, palette='viridis', ax=ax1
)
ax1.set_ylabel('Achievement Rate (%)', fontsize=12, color='b')
ax1.set_title('Target Achievement Rate and Number of Workers by Team', fontsize=14)
ax1.tick_params(axis='y', labelcolor='b')
ax1.set_xlabel('Team', fontsize=12)

# Line plot for number of workers on a secondary axis
ax2 = ax1.twinx()
sns.lineplot(
    x='team', y='no_of_workers', data=team_metrics, color='r', marker='o', ax=ax2
)
ax2.set_ylabel('Number of Workers', fontsize=12, color='r')
ax2.tick_params(axis='y', labelcolor='r')

# Show plot
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
